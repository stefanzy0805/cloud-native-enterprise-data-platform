## 1. list 方法

```python
lst = [3, 1, 4, 1, 5, 9, 2, 6]

# --- 增 ---
lst.append(7)           # 末尾追加一个元素 → [3,1,4,...,6,7]
lst.extend([8,9])       # 末尾追加多个元素（接受任意可迭代）
lst.insert(0, 99)       # 在索引 0 处插入 99

# --- 删 ---
lst.remove(1)           # 删除第一个值为 1 的元素（找不到抛 ValueError）
lst.pop()               # 弹出末尾元素，返回该元素
lst.pop(0)              # 弹出索引 0 的元素（O(n)，慎用）
lst.clear()             # 清空列表（等价于 del lst[:]）

# --- 查 ---
lst.index(4)            # 第一个值为 4 的索引（找不到抛 ValueError）
lst.index(4, 2)         # 从索引 2 开始找
lst.count(1)            # 值 1 出现次数

# --- 排序 ---
lst.sort()                          # 原地升序（就地修改，无返回值）
lst.sort(reverse=True)              # 原地降序
lst.sort(key=lambda x: -x)         # 自定义 key
lst.sort(key=lambda x: (x[1],x[0]))# 元组多字段排序
lst.reverse()                       # 原地反转

# --- 复制 ---
lst.copy()              # 浅拷贝（等价于 lst[:]）
import copy
copy.deepcopy(lst)      # 深拷贝（嵌套结构用）

# --- 列表推导式（最 Pythonic）---
[x**2 for x in range(10) if x % 2 == 0]  # 偶数的平方
[x for row in matrix for x in row]        # 展平二维列表

# --- 常用技巧 ---
flat = sum([[1,2],[3,4],[5]], [])   # 展平：[1,2,3,4,5]
unique = list(dict.fromkeys(lst))   # 去重且保持顺序
chunks = [lst[i:i+3] for i in range(0, len(lst), 3)]  # 按3切分
```

## 2. dict 方法

```python
d = {'name': 'Alice', 'age': 30, 'city': 'NYC'}

# --- 读 ---
d['name']                       # 'Alice'（不存在抛 KeyError）
d.get('name')                   # 'Alice'
d.get('phone', 'N/A')           # 找不到返回默认值 'N/A'
d.keys()                        # dict_keys(['name','age','city'])
d.values()                      # dict_values(['Alice',30,'NYC'])
d.items()                       # dict_items([('name','Alice'),...])  迭代首选

# --- 写 ---
d['email'] = 'a@b.com'          # 添加/更新
d.update({'age': 31, 'phone': '123'})  # 批量更新（也可接受关键字参数）
d.update(age=31, phone='123')   # 同上
d.setdefault('score', 0)        # 若 key 不存在则设为 0；存在则不改变，返回已有值

# --- 删 ---
del d['city']                   # 删除（不存在抛 KeyError）
d.pop('city', None)             # 删除并返回；找不到返回 None（不报错）
d.popitem()                     # 删除并返回最后插入的 (key,value) 对
d.clear()                       # 清空

# --- 构造 ---
dict.fromkeys(['a','b','c'], 0)         # {'a':0,'b':0,'c':0}  初始化相同值
dict(zip(['a','b'],[1,2]))              # {'a':1,'b':2}

# --- 合并（Python 3.9+）---
merged = d1 | d2                # 合并，d2 优先（新语法）
d1 |= d2                        # 原地更新

# --- 字典推导式 ---
{k: v*2 for k,v in d.items() if isinstance(v, int)}  # 过滤并变换
{v: k for k,v in d.items()}    # 反转 key/value

# --- 嵌套 dict 安全读取 ---
data = {'user': {'profile': {'name': 'Alice'}}}
name = data.get('user', {}).get('profile', {}).get('name')  # 'Alice'（不会抛错）
```

## 3. set 方法

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# --- 增删 ---
a.add(5)            # 添加单个元素
a.update([6,7,8])   # 添加多个
a.discard(10)       # 删除（不存在不报错）
a.remove(10)        # 删除（不存在抛 KeyError）
a.pop()             # 随机删除并返回一个元素
a.clear()           # 清空

# --- 集合运算（操作符 or 方法，效果相同）---
a | b               # 并集：{1,2,3,4,5,6}     等价 a.union(b)
a & b               # 交集：{3,4}              等价 a.intersection(b)
a - b               # 差集：{1,2}（在a不在b）  等价 a.difference(b)
a ^ b               # 对称差：{1,2,5,6}        等价 a.symmetric_difference(b)

# --- 子集/超集判断 ---
a.issubset(b)       # a ⊆ b 等价 a <= b
a.issuperset(b)     # a ⊇ b 等价 a >= b
a.isdisjoint(b)     # 无交集则 True

# --- 常用场景 ---
# 去重
unique = list(set([1,1,2,3,3]))    # [1,2,3]（顺序不保证）
# 快速成员检测 O(1) vs list O(n)
big_set = set(range(1_000_000))
999999 in big_set    # 极快
# 找两个列表的公共元素
set(list1) & set(list2)
```

## 4. str 方法

```python
s = '  Hello, World!  '

# --- 清洗 ---
s.strip()               # 'Hello, World!'      去首尾空白
s.lstrip()              # 'Hello, World!  '    去左空白
s.rstrip()              # '  Hello, World!'    去右空白
s.strip('!. ')          # 去除首尾指定字符集（不是子串！）

# --- 大小写 ---
s.lower()               # '  hello, world!  '
s.upper()               # '  HELLO, WORLD!  '
s.title()               # '  Hello, World!  '  每词首字母大写
s.capitalize()          # 仅首字母大写
s.swapcase()            # 大小写互换

# --- 查找 ---
s.find('World')         # 9  找不到返回 -1
s.rfind('l')            # 最后一个 'l' 的位置
s.index('World')        # 9  找不到抛 ValueError
s.count('l')            # 3  统计出现次数
s.startswith('  H')     # True
s.endswith('!  ')       # True
s.startswith(('Hello','Hi'))  # 可传元组

# --- 替换 ---
s.replace('World', 'Python')          # 替换全部
s.replace('l', 'L', 2)               # 最多替换 2 次

# --- 分割 ---
'a,b,,c'.split(',')                   # ['a','b','','c']  按分隔符
'a,b,,c'.split(',', maxsplit=1)       # ['a','b,,c']  最多分1次
'line1\nline2\nline3'.splitlines()   # ['line1','line2','line3']  按行分
'hello'.partition('l')               # ('he','l','lo')  → 找第一个分隔符，三段

# --- 拼接 ---
', '.join(['a','b','c'])              # 'a, b, c'  （推荐，比 + 快）
'\n'.join(lines)                     # 按行合并

# --- 判断 ---
'123'.isdigit()         # True  （纯数字字符）
'abc'.isalpha()         # True  （纯字母）
'abc123'.isalnum()      # True  （字母或数字）
'   '.isspace()         # True  （纯空白）
'Hello'.istitle()       # True  （首字母大写）

# --- 对齐 & 填充 ---
'hi'.center(10)         # '    hi    '
'hi'.ljust(10, '-')     # 'hi--------'
'hi'.rjust(10, '0')     # '00000000hi'
'42'.zfill(5)           # '00042'  数字补零

# --- 格式化 ---
f'Name: {name!r}, Age: {age:03d}'          # f-string（推荐）
f'{3.14159:.2f}'                           # '3.14'
f'{1000000:,}'                             # '1,000,000'
f'{0.875:.1%}'                             # '87.5%'
f'{"left":<10} {"right":>10}'              # 对齐
'{name} is {age}'.format(name='Alice', age=30)  # .format()

# --- 编码 ---
'hello'.encode('utf-8')         # b'hello'
b'hello'.decode('utf-8')        # 'hello'

# --- 其他 ---
s.expandtabs(4)         # 把 \t 展开为空格
s.translate(str.maketrans('aeiou','AEIOU'))  # 字符映射替换
```

## 5. re — 正则表达式

```python
import re

text = 'Order #1234 placed on 2024-01-15 for $99.99'

# re.search(pattern, string)  → 搜索第一个匹配，返回 Match 对象或 None
m = re.search(r'\d+', text)
m.group()       # '1234'
m.start()       # 开始位置
m.end()         # 结束位置
m.span()        # (start, end)

# re.match(pattern, string)  → 只匹配开头（少用）

# re.findall(pattern, string)  → 所有匹配，返回列表
re.findall(r'\d+', text)          # ['1234','2024','01','15','99','99']
re.findall(r'(\d{4})-(\d{2})-(\d{2})', text)  # [('2024','01','15')]

# re.finditer(pattern, string)  → 返回迭代器（省内存）
for m in re.finditer(r'\d+', text):
    print(m.group(), m.start())

# re.sub(pattern, repl, string, count=0)  → 替换
re.sub(r'\d+', 'NUM', text)          # 替换所有数字
re.sub(r'\d+', 'NUM', text, count=1) # 只替换第一个
re.sub(r'(\w+)@(\w+)', r'\2@\1', 'user@host')  # 反向引用

# re.split(pattern, string)  → 按正则分割
re.split(r'[,;\s]+', 'a,b; c  d')    # ['a','b','c','d']

# re.compile  → 预编译，多次使用时更快
pattern = re.compile(r'^\d{4}-\d{2}-\d{2}$')  # 日期格式
pattern.match('2024-01-15')          # Match
pattern.match('not-a-date')          # None

# 常用模式速查
r'\d'        # 数字 [0-9]
r'\D'        # 非数字
r'\w'        # 单词字符 [a-zA-Z0-9_]
r'\s'        # 空白 [ \t\n\r\f\v]
r'\b'        # 单词边界
r'.+'        # 一个或多个任意字符（贪婪）
r'.+?'       # 非贪婪
r'(?P<name>\w+)'  # 命名分组
m.group('name')   # 按名取分组

# flags
re.IGNORECASE   # re.I  大小写不敏感
re.MULTILINE    # re.M  ^ $ 匹配每行
re.DOTALL       # re.S  . 匹配换行符
```
