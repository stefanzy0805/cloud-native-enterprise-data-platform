# 1. 数据类型与变量

````python
# ===== 基本数据类型 =====

# 整数 (int) - 无大小限制
age = 25
big_number = 1_000_000_000  # 下划线分隔，更易读

# 浮点数 (float)
price = 19.99
scientific = 1.5e-10  # 科学计数法

# 布尔值 (bool)
is_active = True
is_deleted = False

# 字符串 (str)
name = "Alice"
message = 'Hello, World!'
multiline = """这是
多行
字符串"""

# None - 表示空值
result = None

print(f"age: {age}, type: {type(age)}")
print(f"price: {price}, type: {type(price)}")
print(f"is_active: {is_active}, type: {type(is_active)}")
print(f"name: {name}, type: {type(name)}")
print(f"result: {result}, type: {type(result)}")
````

````python
# ===== 类型转换 =====

# 转整数
print(int("42"))       # 字符串 -> 整数: 42
print(int(3.7))        # 浮点数 -> 整数: 3 (截断，不是四舍五入)
print(int(True))       # 布尔 -> 整数: 1

# 转浮点数
print(float("3.14"))   # 字符串 -> 浮点: 3.14
print(float(42))       # 整数 -> 浮点: 42.0

# 转字符串
print(str(42))         # 整数 -> 字符串: "42"
print(str([1, 2, 3]))  # 列表 -> 字符串: "[1, 2, 3]"

# 转布尔 (Falsy values: 0, 0.0, "", [], {}, None, False)
print(bool(0))         # False
print(bool(""))        # False
print(bool([]))        # False
print(bool(1))         # True
print(bool("hello"))   # True
````

````python
# ===== 运算符 =====

# 算术运算
print(f"10 + 3 = {10 + 3}")   # 加法: 13
print(f"10 - 3 = {10 - 3}")   # 减法: 7
print(f"10 * 3 = {10 * 3}")   # 乘法: 30
print(f"10 / 3 = {10 / 3}")   # 除法: 3.333... (返回浮点数)
print(f"10 // 3 = {10 // 3}") # 整除: 3
print(f"10 % 3 = {10 % 3}")   # 取余: 1
print(f"10 ** 3 = {10 ** 3}") # 幂运算: 1000

# 比较运算
print(f"5 == 5: {5 == 5}")    # 等于
print(f"5 != 3: {5 != 3}")    # 不等于
print(f"5 > 3: {5 > 3}")      # 大于
print(f"5 >= 5: {5 >= 5}")    # 大于等于

# 逻辑运算
print(f"True and False: {True and False}")  # 与: False
print(f"True or False: {True or False}")    # 或: True
print(f"not True: {not True}")              # 非: False
````

````python
# ===== 赋值技巧 =====

# 多变量赋值
a, b, c = 1, 2, 3
print(f"a={a}, b={b}, c={c}")

# 交换变量 (Python 特有写法)
a, b = b, a
print(f"交换后: a={a}, b={b}")

# 解包
first, *rest = [1, 2, 3, 4, 5]
print(f"first={first}, rest={rest}")  # first=1, rest=[2, 3, 4, 5]

first, *middle, last = [1, 2, 3, 4, 5]
print(f"first={first}, middle={middle}, last={last}")  # first=1, middle=[2, 3, 4], last=5
````

# 2. 字符串操作

DE 经常需要处理各种文本数据，字符串操作必须熟练！

````python
# ===== 字符串基础 =====

s = "Hello, World!"

# 索引 (从 0 开始，支持负索引)
print(f"第一个字符: {s[0]}")     # H
print(f"最后一个字符: {s[-1]}")  # !

# 切片 [start:end:step]
print(f"前5个字符: {s[:5]}")     # Hello
print(f"第7个开始: {s[7:]}")     # World!
print(f"倒数5个: {s[-6:]}")      # World!
print(f"隔一个取: {s[::2]}")     # Hlo ol!
print(f"反转: {s[::-1]}")        # !dlroW ,olleH

# 长度
print(f"长度: {len(s)}")
````

````python
# ===== 常用字符串方法 =====

s = "  Hello, World!  "

# 大小写转换
print(f"upper: {s.upper()}")      # 全大写
print(f"lower: {s.lower()}")      # 全小写
print(f"title: {s.title()}")      # 首字母大写
print(f"capitalize: {s.capitalize()}")  # 句首大写

# 去空白
print(f"strip: '{s.strip()}'")    # 去两边空白
print(f"lstrip: '{s.lstrip()}'")  # 去左边空白
print(f"rstrip: '{s.rstrip()}'")  # 去右边空白

# 查找
s = "Hello, World!"
print(f"find 'World': {s.find('World')}")      # 返回索引 7
print(f"find 'Python': {s.find('Python')}")    # 找不到返回 -1
print(f"'World' in s: {'World' in s}")         # 判断是否存在
print(f"count 'l': {s.count('l')}")            # 计数: 3

# 替换
print(f"replace: {s.replace('World', 'Python')}")
````

````python
# ===== 分割与连接 (DE 常用!) =====

# 分割
csv_line = "Alice,25,Engineer,NYC"
fields = csv_line.split(",")
print(f"split: {fields}")  # ['Alice', '25', 'Engineer', 'NYC']

# 按空白分割 (处理多个空格)
text = "Hello   World  Python"
words = text.split()  # 不传参数，按任意空白分割
print(f"split空白: {words}")  # ['Hello', 'World', 'Python']

# 限制分割次数
log_line = "2023-01-15 10:30:45 ERROR Something went wrong"
parts = log_line.split(" ", 3)  # 最多分割 3 次
print(f"限制分割: {parts}")  # ['2023-01-15', '10:30:45', 'ERROR', 'Something went wrong']

# 按行分割
multiline = "line1\nline2\nline3"
lines = multiline.splitlines()
print(f"splitlines: {lines}")

# 连接
words = ["Hello", "World", "Python"]
print(f"join with space: {' '.join(words)}")
print(f"join with comma: {','.join(words)}")
print(f"join with nothing: {''.join(words)}")
````

# 3. 列表、元组、集合、字典

Python 的四大数据结构，DE 必须精通！

````python
# ===== 列表 (List) - 可变、有序 =====

# 创建
nums = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]  # 可以混合类型
empty = []
from_range = list(range(5))  # [0, 1, 2, 3, 4]

# 访问
print(f"第一个: {nums[0]}")      # 1
print(f"最后一个: {nums[-1]}")   # 5
print(f"切片: {nums[1:4]}")      # [2, 3, 4]

# 修改
nums[0] = 10
print(f"修改后: {nums}")  # [10, 2, 3, 4, 5]
````

````python
# ===== 列表常用方法 =====

nums = [3, 1, 4, 1, 5, 9, 2, 6]

# 添加元素
nums.append(7)          # 末尾添加: [3, 1, 4, 1, 5, 9, 2, 6, 7]
nums.insert(0, 0)       # 指定位置插入: [0, 3, 1, ...]
nums.extend([8, 9])     # 批量添加: [..., 8, 9]
print(f"添加后: {nums}")

# 删除元素
nums = [3, 1, 4, 1, 5]
nums.remove(1)          # 删除第一个值为 1 的元素
print(f"remove(1): {nums}")  # [3, 4, 1, 5]

popped = nums.pop()     # 弹出最后一个
print(f"pop(): {popped}, 剩余: {nums}")  # 5, [3, 4, 1]

popped = nums.pop(0)    # 弹出指定位置
print(f"pop(0): {popped}, 剩余: {nums}")  # 3, [4, 1]

# 排序
nums = [3, 1, 4, 1, 5, 9, 2]
nums.sort()             # 原地排序
print(f"sort(): {nums}")  # [1, 1, 2, 3, 4, 5, 9]

nums.sort(reverse=True) # 降序
print(f"sort(reverse): {nums}")

# sorted() 返回新列表，不修改原列表
nums = [3, 1, 4]
sorted_nums = sorted(nums)
print(f"原列表: {nums}, sorted: {sorted_nums}")

# 反转
nums.reverse()
print(f"reverse(): {nums}")

# 其他
nums = [3, 1, 4, 1, 5]
print(f"index(4): {nums.index(4)}")   # 2 (第一次出现的索引)
print(f"count(1): {nums.count(1)}")   # 2 (出现次数)
print(f"len: {len(nums)}")            # 5
print(f"sum: {sum(nums)}")            # 14
print(f"min: {min(nums)}, max: {max(nums)}")

#隔一个取一个
nums[::2]    # 从索引0开始，每隔一个取一个（偶数索引）
nums[1::2]   # 从索引1开始，每隔一个取一个（奇数索引）
````

````python
# ===== 列表推导式 (非常重要!) =====

# 基础语法: [expression for item in iterable]
squares = [x**2 for x in range(10)]
print(f"平方数: {squares}")

# 带条件: [expression for item in iterable if condition]
even_squares = [x**2 for x in range(10) if x % 2 == 0]
print(f"偶数平方: {even_squares}")

# 嵌套
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print(f"展平: {flattened}")

# if-else (注意位置不同!)
labels = ["even" if x % 2 == 0 else "odd" for x in range(5)]
print(f"标签: {labels}")

# 实际例子: 处理数据
raw_data = ["  Alice  ", "Bob", "  Charlie", None, ""]
clean_data = [s.strip() for s in raw_data if s]  # 过滤空值并去空白
print(f"清洗后: {clean_data}")
````

````python
# ### 练习 5.2: 默认参数与关键字参数
# 编写一个函数 `format_name(first, last, middle=None, title=None, suffix=None)`：
# - 返回格式化的全名
# - 如果有 title 则加在最前面
# - 如果有 middle 则加在 first 和 last 之间
# - 如果有 suffix 则加在最后面

# 例如：`format_name("John", "Doe", title="Dr.", suffix="Jr.")` 返回 `"Dr. John Doe Jr."`


# 练习 5.2: 默认参数与关键字参数
def format_name(first, last, middle=None, title=None, suffix=None):
    # TODO: 格式化全名
    parts = []
    if title is  not None:
        parts.append(title)
    parts.append(first)
    if middle is not None:
        parts.append(middle)
    parts.append(last)
    if suffix is not None:
        parts.append(suffix)
    return " ".join(parts)

# 测试
print(format_name("John", "Doe"))
print(format_name("John", "Doe", title="Dr.", suffix="Jr."))
print(format_name("John", "Doe", middle="William"))
````

````python

# ### 练习 5.4: **kwargs 实战
# 编写一个函数 `build_query(table, **conditions)`，生成简单的 SQL WHERE 子句：
# - `build_query("users", name="Alice", age=25)` 返回 `"SELECT * FROM users WHERE name='Alice' AND age=25"`

# 练习 5.4: **kwargs 实战
def build_query(table, **conditions):
    # TODO: 生成 SQL 查询
    where_clause = []
    for key, value in conditions.items():
        where_clause.append(str(key) + "='" + str(value)+"'")
    all_clause = " AND ".join(where_clause)
    return(f"SELECT * FROM {table} WHERE {all_clause}")

#推荐方案
def build_query(table, **conditions):
    if not conditions:
        return f"SELECT * FROM {table}"

    clauses = []
    for key, value in conditions.items():
        if isinstance(value, str):
            clauses.append(f"{key}='{value}'")
        else:
            clauses.append(f"{key}={value}")

    where_clause = " AND ".join(clauses)
    return f"SELECT * FROM {table} WHERE {where_clause}"
    
# 测试
print(build_query("users", name="Alice", age=25))
````

````python

# ### 练习 5.7: 装饰器基础
# 编写一个装饰器 `@timer`，用于测量函数执行时间并打印。

# 练习 5.7: 装饰器基础
import time

def timer(func):
    def wrapper():
        start = time.time()
        result = func()
        end = time.time()
        print(f"Execution time: {end - start:.4f} seconds")
        return result
    return wrapper


def log(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log
@timer
def slow_function():
    time.sleep(0.5)
    return "done"

slow_function()

````

````python
# ===== 自定义异常 =====

class DataValidationError(Exception):
    """数据验证失败时抛出"""
    pass

class RecordNotFoundError(Exception):
    """记录不存在时抛出"""
    def __init__(self, record_id):
        self.record_id = record_id
        super().__init__(f"Record {record_id} not found")

# 使用
def validate_age(age):
    if not isinstance(age, int):
        raise DataValidationError(f"Age must be int, got {type(age).__name__}")
    if age < 0 or age > 150:
        raise DataValidationError(f"Age {age} is out of valid range")
    return True

try:
    validate_age(-5)
except DataValidationError as e:
    print(f"验证失败: {e}")
````

# 8. 面向对象基础

````python
# ===== 定义类 =====

class Person:
    """表示一个人的类"""
    
    # 类属性 (所有实例共享)
    species = "Human"
    
    def __init__(self, name, age):
        """构造函数"""
        # 实例属性
        self.name = name
        self.age = age
    
    def greet(self):
        """实例方法"""
        return f"Hello, I'm {self.name}, {self.age} years old."
    
    def __str__(self):
        """字符串表示"""
        return f"Person({self.name}, {self.age})"
    
    def __repr__(self):
        """开发者表示"""
        return f"Person(name='{self.name}', age={self.age})"

# 使用
alice = Person("Alice", 25)
print(alice.greet())
print(f"str: {str(alice)}")
print(f"repr: {repr(alice)}")
print(f"species: {alice.species}")
````

````python
# ===== 继承 =====

class Employee(Person):
    """员工类，继承自 Person"""
    
    def __init__(self, name, age, department):
        # 调用父类构造函数
        super().__init__(name, age)
        self.department = department
    
    def greet(self):
        """重写父类方法"""
        return f"Hello, I'm {self.name} from {self.department}."
    
    def __str__(self):
        return f"Employee({self.name}, {self.department})"

# 使用
bob = Employee("Bob", 30, "Engineering")
print(bob.greet())
print(bob.age)  # 继承自 Person
````

````python
# ===== 类方法和静态方法 =====

class DateUtils:
    date_format = "%Y-%m-%d"
    
    @classmethod
    def set_format(cls, fmt):
        """类方法: 可以访问类属性"""
        cls.date_format = fmt
    
    @staticmethod
    def is_valid_date(date_str):
        """静态方法: 不访问类或实例"""
        from datetime import datetime
        try:
            datetime.strptime(date_str, "%Y-%m-%d")
            return True
        except ValueError:
            return False

print(f"Valid date: {DateUtils.is_valid_date('2023-01-15')}")
print(f"Invalid date: {DateUtils.is_valid_date('2023-13-01')}")
````

````python
# ===== @property 装饰器 =====

class Circle:
    def __init__(self, radius):
        self._radius = radius
    
    @property
    def radius(self):
        """Getter"""
        return self._radius
    
    @radius.setter
    def radius(self, value):
        """Setter with validation"""
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value
    
    @property
    def area(self):
        """计算属性 (只读)"""
        import math
        return math.pi * self._radius ** 2

c = Circle(5)
print(f"Radius: {c.radius}, Area: {c.area:.2f}")

c.radius = 10
print(f"New radius: {c.radius}, New area: {c.area:.2f}")
````

````python
# ===== dataclass (Python 3.7+, 推荐!) =====

from dataclasses import dataclass, field
from typing import List

@dataclass
class DataRecord:
    """数据记录类"""
    id: int
    name: str
    value: float
    tags: List[str] = field(default_factory=list)
    
    def total_value(self, multiplier: float = 1.0) -> float:
        return self.value * multiplier

# 自动生成 __init__, __repr__, __eq__ 等
record = DataRecord(1, "test", 100.0, ["a", "b"])
print(record)
print(f"Total: {record.total_value(1.1)}")

# 比较
record2 = DataRecord(1, "test", 100.0, ["a", "b"])
print(f"Equal: {record == record2}")
````

````python
# ===== map, filter, reduce =====

nums = [1, 2, 3, 4, 5]

# map: 对每个元素应用函数
squares = list(map(lambda x: x**2, nums))
print(f"map: {squares}")

# filter: 过滤元素
evens = list(filter(lambda x: x % 2 == 0, nums))
print(f"filter: {evens}")

# reduce: 累积计算
from functools import reduce
product = reduce(lambda x, y: x * y, nums)
print(f"reduce (乘积): {product}")

# 实际上，推导式通常更 Pythonic
squares = [x**2 for x in nums]
evens = [x for x in nums if x % 2 == 0]
````

````python
# ===== enumerate, zip =====

# enumerate: 带索引遍历
fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")

# zip: 并行遍历
names = ["Alice", "Bob"]
ages = [25, 30]
for name, age in zip(names, ages):
    print(f"{name} is {age}")

# zip 创建字典
person = dict(zip(["name", "age"], ["Alice", 25]))
print(f"dict from zip: {person}")
````

````python
# ===== datetime: 日期时间处理 =====

from datetime import datetime, date, timedelta

# 当前时间
now = datetime.now()
today = date.today()
print(f"Now: {now}")
print(f"Today: {today}")

# 创建日期
d = datetime(2023, 1, 15, 10, 30, 0)
print(f"Created: {d}")

# 格式化
formatted = now.strftime("%Y-%m-%d %H:%M:%S")
print(f"Formatted: {formatted}")

# 解析
parsed = datetime.strptime("2023-01-15 10:30:00", "%Y-%m-%d %H:%M:%S")
print(f"Parsed: {parsed}")

# 日期运算
tomorrow = today + timedelta(days=1)
last_week = today - timedelta(weeks=1)
print(f"Tomorrow: {tomorrow}, Last week: {last_week}")

# 日期差
diff = datetime(2023, 12, 31) - datetime(2023, 1, 1)
print(f"Days in 2023: {diff.days}")
````

````python
# ===== re: 正则表达式 =====

import re

text = "Contact us at support@example.com or sales@example.com"

# 查找所有邮箱
emails = re.findall(r'[\w.]+@[\w.]+', text)
print(f"Emails: {emails}")

# 替换
censored = re.sub(r'[\w.]+@', '***@', text)
print(f"Censored: {censored}")

# 匹配
pattern = r'^\d{4}-\d{2}-\d{2}$'
print(f"Is date: {bool(re.match(pattern, '2023-01-15'))}")
print(f"Is date: {bool(re.match(pattern, 'not-a-date'))}")

# 分组
log = "2023-01-15 10:30:45 ERROR Something went wrong"
match = re.match(r'(\d{4}-\d{2}-\d{2}) (\d{2}:\d{2}:\d{2}) (\w+) (.+)', log)
if match:
    date, time, level, message = match.groups()
    print(f"Date: {date}, Time: {time}, Level: {level}, Message: {message}")
````
