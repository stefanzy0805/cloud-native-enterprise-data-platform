```python
# ===== 生成器表达式 vs 列表推导式 =====
import sys

# 列表: 立即计算，占用内存
list_comp = [x**2 for x in range(1000000)]
print(f"列表大小: {sys.getsizeof(list_comp):,} bytes")

# 生成器: 惰性计算，几乎不占内存
gen_exp = (x**2 for x in range(1000000))
print(f"生成器大小: {sys.getsizeof(gen_exp):,} bytes")
```

```python
# ===== yield 关键字 =====
def read_large_file(file_path, chunk_size=1000):
    """模拟分块读取大文件 - DE 常用模式"""
    with open(file_path, 'r') as f:
        while True:
            lines = f.readlines(chunk_size)
            if not lines:
                break
            yield lines  # 每次只返回一块，不占用全部内存

# 使用示例 (伪代码)
# for chunk in read_large_file('huge_file.csv'):
#     process(chunk)
```

```python
# ===== 实用生成器: 批量处理 =====
def batch_generator(items, batch_size):
    """将数据分批 - 写入数据库/API调用时常用"""
    batch = []
    for item in items:
        batch.append(item)
        if len(batch) == batch_size:
            yield batch
            batch = []
    if batch:  # 别忘了最后一批
        yield batch

# 示例
data = list(range(25))
for i, batch in enumerate(batch_generator(data, 10)):
    print(f"Batch {i}: {batch}")
```

```python
# ===== 装饰器: 日志记录 (DE 必备) =====
import functools
import time

def log_execution_time(func):
    """记录函数执行时间的装饰器"""
    @functools.wraps(func)  # 保留原函数的元信息
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        elapsed = time.time() - start
        print(f"[{func.__name__}] 执行时间: {elapsed:.3f}s")
        return result
    return wrapper

@log_execution_time
def slow_function():
    time.sleep(1)
    return "done"

slow_function()
```

```python
# ===== 装饰器: 重试机制 (调用 API/数据库必备) =====
import random

def retry(max_attempts=3, delay=1):
    """带参数的装饰器: 失败自动重试"""
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(1, max_attempts + 1):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    print(f"Attempt {attempt} failed: {e}")
                    if attempt < max_attempts:
                        time.sleep(delay)
                    else:
                        raise
        return wrapper
    return decorator

@retry(max_attempts=3, delay=0.5)
def unreliable_api_call():
    """模拟不稳定的 API"""
    if random.random() < 0.7:  # 70% 失败率
        raise Exception("Connection timeout")
    return {"status": "success"}

try:
    result = unreliable_api_call()
    print(f"成功: {result}")
except Exception as e:
    print(f"最终失败: {e}")
```

## 1.4 🟢 Context Manager (上下文管理器)

管理资源（文件、数据库连接、锁）的标准方式。

```python
# ===== 使用 contextlib =====
from contextlib import contextmanager

@contextmanager
def database_connection(connection_string):
    """模拟数据库连接的上下文管理器"""
    print(f"Connecting to {connection_string}...")
    conn = {"status": "connected", "db": connection_string}  # 模拟连接对象
    try:
        yield conn  # 把连接交给 with 块使用
    except Exception as e:
        print(f"Error occurred: {e}")
        raise
    finally:
        print("Closing connection...")  # 无论如何都会执行

# 使用
with database_connection("postgresql://localhost/mydb") as conn:
    print(f"Working with: {conn}")
    # 如果这里抛异常，连接也会被正确关闭
```

```python
# ===== 实用: 计时器上下文管理器 =====
@contextmanager
def timer(description="Operation"):
    """计时器 - 方便测量代码块执行时间"""
    start = time.time()
    yield
    elapsed = time.time() - start
    print(f"{description} took {elapsed:.3f}s")

with timer("Data processing"):
    # 模拟耗时操作
    time.sleep(0.5)
    data = [i**2 for i in range(100000)]
```

## 1.5 🟢 异常处理

生产代码必须有健壮的异常处理。

```python
# ===== 完整的异常处理模式 =====
class DataValidationError(Exception):
    """自定义异常: 数据验证失败"""
    pass

def process_record(record):
    """处理单条记录"""
    try:
        # 数据验证
        if not record.get("id"):
            raise DataValidationError("Missing required field: id")
        
        # 类型转换
        amount = float(record["amount"])
        
        # 业务逻辑
        return {"id": record["id"], "processed_amount": amount * 1.1}
    
    except DataValidationError:
        raise  # 验证错误直接向上抛
    except (KeyError, TypeError) as e:
        raise DataValidationError(f"Invalid record format: {e}")
    except ValueError as e:
        raise DataValidationError(f"Invalid amount value: {e}")

# 测试
good_record = {"id": 1, "amount": "100.50"}
bad_record = {"amount": "not a number"}

print(process_record(good_record))
try:
    process_record(bad_record)
except DataValidationError as e:
    print(f"验证失败: {e}")
```

## 1.6 🟡 collections 模块

```python
from collections import Counter, defaultdict, namedtuple, deque

# ===== Counter: 计数神器 =====
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
word_counts = Counter(words)
print(f"词频统计: {word_counts}")
print(f"最常见的2个: {word_counts.most_common(2)}")

# 两个 Counter 可以相加
more_words = Counter(["apple", "date"])
combined = word_counts + more_words
print(f"合并后: {combined}")
```

```python
# ===== defaultdict: 自动初始化 =====
# 传统方式 - 需要检查 key 是否存在
regular_dict = {}
for word in words:
    if word not in regular_dict:
        regular_dict[word] = 0
    regular_dict[word] += 1

# 用 defaultdict - 更简洁
auto_dict = defaultdict(int)  # 默认值为 0
for word in words:
    auto_dict[word] += 1

print(f"defaultdict 结果: {dict(auto_dict)}")

# 分组场景: 默认值为列表
students = [
    ("Alice", "Math"),
    ("Bob", "Science"),
    ("Alice", "Science"),
    ("Charlie", "Math"),
]

by_subject = defaultdict(list)
for student, subject in students:
    by_subject[subject].append(student)

print(f"按科目分组: {dict(by_subject)}")
```

## 1.8 🟡 Type Hints (类型注解)

现代 Python 项目的标配，提高代码可读性和 IDE 支持。

```python
from typing import List, Dict, Optional, Tuple, Union, Callable
from dataclasses import dataclass

# ===== 基础类型注解 =====
def process_batch(
    records: List[Dict[str, any]],
    batch_size: int = 100,
    callback: Optional[Callable[[int], None]] = None
) -> Tuple[int, int]:
    """
    处理一批记录
    
    Args:
        records: 要处理的记录列表
        batch_size: 每批大小
        callback: 可选的进度回调函数
    
    Returns:
        (成功数, 失败数)
    """
    success = len(records)
    failed = 0
    if callback:
        callback(success)
    return success, failed
```

```python
# ===== dataclass: 现代数据类 (比 namedtuple 更强大) =====
@dataclass
class TaxiTripRecord:
    trip_id: int
    pickup_time: str
    dropoff_time: str
    distance: float
    fare: float
    tip: float = 0.0  # 默认值
    
    @property
    def total(self) -> float:
        return self.fare + self.tip
    
    def tip_percentage(self) -> float:
        return (self.tip / self.fare * 100) if self.fare > 0 else 0

# 使用
trip = TaxiTripRecord(
    trip_id=1,
    pickup_time="2023-01-01 10:00",
    dropoff_time="2023-01-01 10:30",
    distance=5.5,
    fare=18.50,
    tip=4.00
)

print(f"行程: {trip}")
print(f"总费用: ${trip.total:.2f}")
print(f"小费率: {trip.tip_percentage():.1f}%")
```

## 2.1 🟢 窗口函数 (Window Functions)

**面试必考！** 必须熟练掌握 ROW_NUMBER, RANK, DENSE_RANK, LAG, LEAD, SUM OVER 等。

```python
# ===== ROW_NUMBER vs RANK vs DENSE_RANK =====
spark.sql("""
    SELECT 
        payment_type,
        total_amount,
        ROW_NUMBER() OVER (PARTITION BY payment_type ORDER BY total_amount DESC) as row_num,
        RANK() OVER (PARTITION BY payment_type ORDER BY total_amount DESC) as rank,
        DENSE_RANK() OVER (PARTITION BY payment_type ORDER BY total_amount DESC) as dense_rank
    FROM trips
    WHERE payment_type IN (1, 2)
""").filter("row_num <= 5").orderBy("payment_type", "row_num").show(10)

# ROW_NUMBER: 1, 2, 3, 4, 5 (总是连续，即使值相同)
# RANK:       1, 1, 3, 4, 5 (相同值同排名，跳过)
# DENSE_RANK: 1, 1, 2, 3, 4 (相同值同排名，不跳过)
```

```python
# ===== LAG / LEAD: 访问前一行/后一行 =====
# 典型场景: 计算环比变化
spark.sql("""
    WITH daily_revenue AS (
        SELECT 
            DATE(tpep_pickup_datetime) as date,
            ROUND(SUM(total_amount), 2) as revenue
        FROM trips
        GROUP BY DATE(tpep_pickup_datetime)
    )
    SELECT 
        date,
        revenue,
        LAG(revenue, 1) OVER (ORDER BY date) as prev_day_revenue,
        ROUND(
            (revenue - LAG(revenue, 1) OVER (ORDER BY date)) / 
            LAG(revenue, 1) OVER (ORDER BY date) * 100, 
            2
        ) as day_over_day_pct
    FROM daily_revenue
    ORDER BY date
    LIMIT 10
""").show()
```

```python
# ===== SUM / AVG OVER: 累计求和、移动平均 =====
spark.sql("""
    WITH daily_trips AS (
        SELECT 
            DATE(tpep_pickup_datetime) as date,
            COUNT(*) as trip_count
        FROM trips
        GROUP BY DATE(tpep_pickup_datetime)
    )
    SELECT 
        date,
        trip_count,
        -- 累计求和
        SUM(trip_count) OVER (ORDER BY date) as cumulative_trips,
        -- 7天移动平均
        ROUND(AVG(trip_count) OVER (
            ORDER BY date 
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        ), 0) as moving_avg_7d
    FROM daily_trips
    ORDER BY date
    LIMIT 15
""").show()
```

## 2.2 🟢 CTE (Common Table Expressions)

让复杂查询更可读，是写长 SQL 的标准方式。

```python
# ===== 多个 CTE 链式使用 =====
spark.sql("""
    -- 第一步: 计算每天的基础统计
    WITH daily_stats AS (
        SELECT 
            DATE(tpep_pickup_datetime) as date,
            COUNT(*) as trip_count,
            SUM(total_amount) as revenue,
            AVG(trip_distance) as avg_distance
        FROM trips
        GROUP BY DATE(tpep_pickup_datetime)
    ),
    -- 第二步: 添加周信息
    weekly_context AS (
        SELECT 
            *,
            WEEKOFYEAR(date) as week_num,
            DAYOFWEEK(date) as day_of_week
        FROM daily_stats
    ),
    -- 第三步: 计算周平均作为基准
    weekly_avg AS (
        SELECT 
            week_num,
            AVG(trip_count) as weekly_avg_trips
        FROM weekly_context
        GROUP BY week_num
    )
    -- 最终: 比较每天与周平均
    SELECT 
        wc.date,
        wc.day_of_week,
        wc.trip_count,
        ROUND(wa.weekly_avg_trips, 0) as weekly_avg,
        ROUND((wc.trip_count - wa.weekly_avg_trips) / wa.weekly_avg_trips * 100, 1) as pct_vs_avg
    FROM weekly_context wc
    JOIN weekly_avg wa ON wc.week_num = wa.week_num
    ORDER BY wc.date
    LIMIT 14
""").show()
```

```python
# ===== INNER vs LEFT vs FULL OUTER JOIN =====
# 分析机场接送行程
spark.sql("""
    SELECT 
        l.zone_name as pickup_zone,
        p.payment_name,
        COUNT(*) as trip_count,
        ROUND(AVG(t.total_amount), 2) as avg_fare
    FROM trips t
    LEFT JOIN locations l ON t.PULocationID = l.location_id
    LEFT JOIN payment_types p ON t.payment_type = p.payment_type
    WHERE l.zone_name IS NOT NULL  -- 只看有映射的地点
    GROUP BY l.zone_name, p.payment_name
    ORDER BY trip_count DESC
""").show()
```

```python
# ===== Self JOIN: 同一表自连接 =====
# 场景: 找出往返行程 (上车地点 = 另一行程的下车地点)
spark.sql("""
    WITH sample_trips AS (
        SELECT * FROM trips LIMIT 10000
    )
    SELECT 
        t1.PULocationID as start_location,
        t1.DOLocationID as mid_location,
        t2.DOLocationID as end_location,
        t1.total_amount + t2.total_amount as total_fare
    FROM sample_trips t1
    JOIN sample_trips t2 
        ON t1.DOLocationID = t2.PULocationID  -- 下车地点 = 下一程上车地点
        AND t1.PULocationID = t2.DOLocationID  -- 形成往返
        AND t2.tpep_pickup_datetime > t1.tpep_dropoff_datetime  -- 时间顺序正确
    LIMIT 10
""").show()
```

## 2.4 🟢 子查询 vs CTE vs JOIN

三种方式实现相同逻辑，了解各自优缺点。

```python
# 需求: 找出高于平均费用的行程

# ===== 方式1: 子查询 (简单但可能效率低) =====
spark.sql("""
    SELECT COUNT(*) as above_avg_trips
    FROM trips
    WHERE total_amount > (
        SELECT AVG(total_amount) FROM trips
    )
""").show()
```

```python
# ===== 方式2: CTE (更可读) =====
spark.sql("""
    WITH avg_fare AS (
        SELECT AVG(total_amount) as avg_amount FROM trips
    )
    SELECT COUNT(*) as above_avg_trips
    FROM trips, avg_fare
    WHERE trips.total_amount > avg_fare.avg_amount
""").show()
```

```python
# ===== 方式3: 窗口函数 (最灵活) =====
spark.sql("""
    SELECT COUNT(*) as above_avg_trips
    FROM (
        SELECT 
            total_amount,
            AVG(total_amount) OVER () as avg_amount
        FROM trips
    ) t
    WHERE total_amount > avg_amount
""").show()
```

```python
# ===== 参考答案 (先自己尝试！) =====
spark.sql("""
    WITH trip_with_period AS (
        SELECT 
            *,
            CASE 
                WHEN HOUR(tpep_pickup_datetime) BETWEEN 7 AND 9 THEN '早高峰 (7-9)'
                WHEN HOUR(tpep_pickup_datetime) BETWEEN 17 AND 19 THEN '晚高峰 (17-19)'
                WHEN HOUR(tpep_pickup_datetime) >= 22 OR HOUR(tpep_pickup_datetime) <= 5 THEN '深夜 (22-5)'
                ELSE '其他时段'
            END as time_period,
            (UNIX_TIMESTAMP(tpep_dropoff_datetime) - UNIX_TIMESTAMP(tpep_pickup_datetime)) / 3600.0 as trip_hours
        FROM trips
        WHERE trip_distance > 0 
          AND total_amount > 0
          AND tpep_dropoff_datetime > tpep_pickup_datetime
    ),
    overall_avg AS (
        SELECT 
            AVG(trip_distance) as overall_avg_distance,
            AVG(total_amount) as overall_avg_fare
        FROM trip_with_period
    )
    SELECT 
        t.time_period,
        COUNT(*) as trip_count,
        ROUND(AVG(t.trip_distance), 2) as avg_distance,
        ROUND(AVG(t.total_amount), 2) as avg_fare,
        ROUND(AVG(CASE WHEN t.trip_hours > 0 THEN t.trip_distance / t.trip_hours END), 1) as avg_speed_mph,
        ROUND(SUM(CASE WHEN t.payment_type = 1 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) as credit_card_pct,
        ROUND((AVG(t.trip_distance) - o.overall_avg_distance) / o.overall_avg_distance * 100, 1) as distance_vs_avg_pct,
        ROUND((AVG(t.total_amount) - o.overall_avg_fare) / o.overall_avg_fare * 100, 1) as fare_vs_avg_pct
    FROM trip_with_period t
    CROSS JOIN overall_avg o
    GROUP BY t.time_period, o.overall_avg_distance, o.overall_avg_fare
    ORDER BY trip_count DESC
""").show(truncate=False)
```
